```python
import numpy as np
```

```python
#data
x=np.array([1,2,3,4,5],dtype=float)
y=np.array([2,4,5,4,5],dtype=float)
```

```python
#initial values
b0=0
b1=0
learning_rate=0.01
epochs=1000
n=len(x)
```

```python
#gradient descent
for i in range(epochs):1
```

```python
#predictions
y_pred=b0+b1*x
```

```python
#calculate gradients
db0=(-2/n)*np.sum(y-y_pred)
db1=(-2/n)*np.sum(x*(y-y_pred))
```

```python
#upload b0 and b1
b0=b0-learning_rate*db0
b1=b1-learning_rate*db1
print("b0=",b0)
print("b1=",b1)
print("regression equation:")
print("y=",b0,"+",b1,"x")
```
